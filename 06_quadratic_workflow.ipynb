```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict, Literal
```

```python
class QuadraticState(TypedDict):
    a : int
    b : int
    c : int

    
    discriminant : int
    result : str

def show_equation(state : QuadraticState):
    equation = f'{state['a']}*x^2+ {state['b']}*x + {state['c']}'
    return {'equation':equation}

def calculate_discriminant(state : QuadraticState):
    discriminant = state['b']*state['b'] - 4*state['a']*state['c']
    return {'discriminant':discriminant}

def real_equal_roots(state : QuadraticState):
    return {'result' : "Real Equal roots"}

def real_unequal_roots(state : QuadraticState):
    return {'result' : "Real Unequal roots"}

def unreal_roots(state : QuadraticState):
    return {'result' : "Unreal roots"}

def check_condition(state : QuadraticState) -> Literal['real_equal_roots','real_unequal_roots','unreal_roots']:
    if(state['discriminant']==0):
        return 'real_equal_roots'
    elif(state['discriminant']>0):
        return 'real_unequal_roots'
    else:
        return 'unreal_roots'
```

```python
graph = StateGraph(QuadraticState)

graph.add_node('show_equation',show_equation)
graph.add_node('calculate_discriminant',calculate_discriminant)
graph.add_node('real_equal_roots',real_equal_roots)
graph.add_node('real_unequal_roots',real_unequal_roots)
graph.add_node('unreal_roots',unreal_roots)

graph.add_edge(START,'show_equation')
graph.add_edge('show_equation','calculate_discriminant')
graph.add_conditional_edges('calculate_discriminant',check_condition)
graph.add_edge('real_equal_roots',END)
graph.add_edge('real_unequal_roots',END)
graph.add_edge('unreal_roots',END)

workflow = graph.compile()
```

```python
initial_state = {
    'a':1,
    'b':-5,
    'c':6
    }
workflow.invoke(initial_state)
```
